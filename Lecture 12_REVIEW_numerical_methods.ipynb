### d. ODE IVP
Solve dy/dx = 3x + y^2 with y(0) = 1 using RK4 method to estimate y(0.2).

```python
# Your code here
# runge-kutta 4th order method for solving ODEs
def f(x, y):
    return 3*x + y**2

x0, y0 = 0, 1

def runge_kutta_4th_order(f, x0, y0, h, n):
    x, y = x0, y0
    for _ in range(n):
        k1 = h * f(x, y)
        k2 = h * f(x + h/2, y + k1/2)
        k3 = h * f(x + h/2, y + k2/2)
        k4 = h * f(x + h, y + k3)
        y += (k1 + 2*k2 + 2*k3 + k4) / 6
        x += h
    return x, y

print(runge_kutta_4th_order(f, x0, y0, h=0.1, n=10))
```
