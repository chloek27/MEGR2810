## Trapezoidal Rule

### Algorithm:
1. Divide the interval [a, b] into n subintervals of equal width h = (b - a)/n.
2. Compute the sum: Integral ≈ (h/2) * [f(x₀) + 2f(x₁) + 2f(x₂) + ... + 2f(xₙ₋₁) + f(xₙ)].
3. Return the result.


```python
# Your code here
import numpy as np
import matplotlib.pyplot as plt

a, b = [0, 1]
n = 10
x = np.linspace(a, b, n+1)

def trap_rule(f, a, b, n):
    h = (b - a) / n
    integral = 0.5 * (f(a) + f(b))
    for i in range(1, n):
        integral += 2 * f(a + i * h)
    return integral

print("Trapezoidal Rule Integral:", trap_rule)
print("Exact Integral:", integral)
```
