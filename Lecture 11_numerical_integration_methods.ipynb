## Trapezoidal Rule

### Algorithm:
1. Divide the interval [a, b] into n subintervals of equal width h = (b - a)/n.
2. Compute the sum: Integral ≈ (h/2) * [f(x₀) + 2f(x₁) + 2f(x₂) + ... + 2f(xₙ₋₁) + f(xₙ)].
3. Return the result.


```python
# Your code here
import numpy as np
import matplotlib.pyplot as plt

a, b = [0, 1]
n = 10
x = np.linspace(a, b, n+1)

def f(x):
    return np.exp(x)

def trap_rule(f, a, b, n):
    h = (b - a) / n
    integral = 0.5 * (f(a) + f(b))
    for i in range(1, n):
        integral += f(a + i * h)
    return h * integral

approximation = trap_rule(f, a, b, n)
exact = np.exp(1) - np.exp(0)
error = abs(exact - approximation)

print("Trapezoidal Rule Integral:", approximation)
print(f"Exact Integral = {exact}")
print(f"Error = {error}")
```


## Simpson's Rule

### Algorithm:
1. Divide the interval [a, b] into an even number n of subintervals.
2. Compute the sum: Integral ≈ (h/3) * [f(x₀) + 4f(x₁) + 2f(x₂) + 4f(x₃) + ... + 4f(xₙ₋₁) + f(xₙ)].
3. Return the result.


```python
# Your code here
#parabola to est area underneath.
# It's the coefficient of x^2
# Even number * 2, off number * 2 and add thse to their piles of numbers.
import numpy as np
import matplotlib.pyplot as plt

a, b = [0, 1]
n = 10
x = np.linspace(a, b, n+1)

def simp_rule(f, a, b, n):
    if n % 2:
        raise ValueError("n must be even for Simpson's rule")
    else:
        h = (b - a) / n
        integral = f(a) + f(b)
        odd_sum = sum(4 * f(a + i * h) for i in range(1, n, 2))
        even_sum = sum(2 * f(a + i * h) for i in range(2, n-1, 2))
        integral += odd_sum + even_sum
        integral *= h / 3
        return integral
    
f = lambda x: x**2  # parabola
result = simp_rule(f, a, b, n)
print("Estimated integral:", result)
```
