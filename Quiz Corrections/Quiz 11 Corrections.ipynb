```python
# Question 1: Write Python code to use the Trapezoidal Rule to integrate exp (-x^2) from x=0 to x=1.

import numpy as np
# define the functin=on and interval
a, b = 0, 1
n = 10 # nummber of subintervals for trapezoidal rule
x = np.linspace(a, b, n+1)
# Your code below

def f(x): #define the function given
    return np.exp(-x**2)

def trap_rule(f, a, b, n): # defining trapezoidal rule function
    h = (b - a) / n # subinterval
    integral = 0.5 * (f(a) + f(b)) # computing function values at the endpoints
    for i in range(1, n): # for loop to sum the function values at interior points
        integral += f(a + i * h)
    return integral *h # multiply by the width of subintervals

trap_integral = trap_rule(f, a, b, n) # assigning the result to a variable

print(f"Trapezoidal Rule Approximation: {trap_integral:.6f}") # printing the variable's answer.
```
